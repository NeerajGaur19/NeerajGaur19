**2- Prompt Template**

- reusable , parameterized instructions

```python
from langchain_core.prompts import PromptTemplate
```

```python
template1 = PromptTemplate(
    template = 'Write a detailed report on {topic}',
    input_variables = ['topic']
)

template2 = PromptTemplate(
    template = 'Write a 5 line summary on /n {topic}',
    input_variables = ['topic']
)

prompt1 = template1.invoke({"topic":"attention is all you need"})
print(model.invoke(prompt1).content)
prompt2 = template2.invoke({"topic":"harrypotter"})
print(model.invoke(prompt2).content)
```

**3- Memory** - To store or rember past conversation

- LangChain stores conversation as messages --------

  - HumanMessage - user input

  - AIMessage - response generate by AI

  - SystemMessage - Instructions

```python
from langchain_core.messages  import HumanMessage, AIMessage , SystemMessage
```

```python
message = [
    SystemMessage(content = 'You are my first chatbot'),
    HumanMessage(content='Tell me something about Virat Kohli')
]

result = model.invoke(message)
message.append(AIMessage(content = result.content))
```

```python
print(message)
```

```python
# CHATBOT WITH MEMORY ------------------------------------------------------------------
# ----------------------------------------------------------------------------------------------
chat_history = [
    SystemMessage(content='You are a helpful Assistance')
]

while True:
  user = input('You: ')
  chat_history.append(HumanMessage(content = user))
  if user =="exit":
    break
  result = model.invoke(chat_history)
  chat_history.append(AIMessage(content = result.content))
  print('AI: ',result.content)
```

**4- OutputParser**

- An Output Parser takes the response generated by the LLM and convert it into a format that your python program can easily use

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
template1 = PromptTemplate(
    template = 'Write a detailed report on {topic}',
    input_variables = ['topic']
)

template2 = PromptTemplate(
    template = 'Write a 5 line summary on /n {text}',
    input_variables = ['text']
)

parser = StrOutputParser()
Chain = template1| model | parser
result = Chain.invoke({'topic': 'black hole'})
result
```

**5- Chain**

- A chain is a sequence of LangChain components connected together

- The output of one component automatically becomes the input of the next component

- **Runnable chain** - using | operator

```python
prompt = PromptTemplate(
    template = ' Generate 5 interesting facts  about {topic}',
    input_variables = ['topic']
)

parser = StrOutputParser()
chain = prompt | model | parser

result = chain.invoke({'topic':'cricket'})
print(result)
```
